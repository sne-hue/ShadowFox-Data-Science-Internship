1,629 films. 2001 to 2019. Every genre from Biography to Horror, every rating from a brutal 1.7 to a
near-perfect 9.4. Somewhere in this pile of numbers is the answer to a question every movie-goer
argues about at some point: **does star power actually mean a good movie? Does a longer runtime mean
a better one? And which genre is India's most reliable crowd-pleaser?**

| # | Question | Headline Answer |
|---|---|---|
| 1 | How has the genre mix shifted since 2001? | Drama never lost its crown — Romance quietly lost half its audience |
| 2 | Does popularity mean quality? | A loose relationship (0.44) — hype helps, but doesn't guarantee it |
| 3 | Which genre rates highest? | Biography (6.58) crushes Horror (4.70) |
| 4 | Does runtime matter? | Barely at all (0.02) — nobody's rating a film by the clock |
| 5 | Has quality changed over time? | A real uptick from 2017 onward |...
.and soo on


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl
import seaborn as sns
import re

%matplotlib inline
PALETTE = ["#E63946", "#F1A208", "#2A9D8F", "#457B9D", "#8E44AD", "#1D3557"]
sns.set_theme(style="whitegrid", palette=PALETTE)
mpl.rcParams.update({
    "figure.figsize": (9, 4.8),
    "axes.titlesize": 14,
    "axes.titleweight": "bold",
    "axes.edgecolor": "#333333",
    "font.size": 11,
})

df = pd.read_csv(r"C:\Users\User\Downloads\movies.csv")
print(f"Rows: {df.shape[0]:,}  |  Columns: {df.shape[1]}")
df.head(3)
```

```python
df["runtime"] = pd.to_numeric(df["runtime"], errors="coerce")
df["genres_list"] = df["genres"].str.split("|")

def parse_wins(s):
    if pd.isna(s):
        return 0
    m = re.search(r"(\d+)\s*win", s)
    return int(m.group(1)) if m else 0

def parse_noms(s):
    if pd.isna(s):
        return 0
    m = re.search(r"(\d+)\s*nomination", s)
    return int(m.group(1)) if m else 0

df["wins"] = df["wins_nominations"].apply(parse_wins)
df["nominations"] = df["wins_nominations"].apply(parse_noms)
df["total_recognition"] = df["wins"] + df["nominations"]
df["decade"] = (df["year_of_release"] // 10) * 10

df[["title_x", "runtime", "genres_list", "wins", "nominations", "imdb_rating"]].head()
```

##  Question 1: What Does India Actually Watch?

```python
exploded = df.explode("genres_list")
genre_counts = exploded["genres_list"].value_counts()

plt.figure(figsize=(9, 5))
bars = plt.bar(genre_counts.head(12).index, genre_counts.head(12).values,
                color=PALETTE[5], edgecolor="black")
bars[0].set_color(PALETTE[0])
plt.title("Drama Wins ")
plt.ylabel("Number of Films")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

**Drama shows up in 67% of every film in this dataset** — 1,085 out of 1,629. Comedy (571), Romance (451), and Action (431) fight it out for a
distant second place.

```python
top4 = ["Drama", "Comedy", "Romance", "Action"]
trend = exploded[exploded["genres_list"].isin(top4)].groupby(
    ["year_of_release", "genres_list"]
).size().unstack(fill_value=0)

trend.plot(figsize=(10, 5), marker="o", linewidth=2.2, color=PALETTE[:4])
plt.title("Romance Had a Quiet Breakup With Bollywood")
plt.ylabel("Number of Films")
plt.xlabel("Year")
plt.legend(title="Genre")
plt.show()
```

 **Romance roughly halved** — from a peak of 33-34
films a year in the mid-2000s down to about 15-18 by 2018-2019. Drama, Comedy, and Action all stayed
remarkably stable across two decades. Somewhere around 2010, Indian cinema quietly fell a little out
of love with the love story.

##  Question 2: Does Popularity Actually Mean Quality?

```python
df["log_votes"] = np.log1p(df["imdb_votes"])

plt.figure(figsize=(7.5, 6))
plt.scatter(df["log_votes"], df["imdb_rating"], alpha=0.35, color=PALETTE[3], edgecolor="none")
corr = df["log_votes"].corr(df["imdb_rating"])
plt.xlabel("log(1 + IMDb votes)  —  a crowd-size meter")
plt.ylabel("IMDb Rating")
plt.title(f"Hype vs. Quality (r = {corr:.2f})")
plt.show()
```

## Question 3: Which Genre Actually Delivers?

```python
genre_ratings = exploded.groupby("genres_list")["imdb_rating"].agg(["mean", "count"])
genre_ratings = genre_ratings[genre_ratings["count"] >= 50].sort_values("mean", ascending=False)

plt.figure(figsize=(9, 5.5))
colors = [PALETTE[2] if v == genre_ratings["mean"].max() else
          PALETTE[0] if v == genre_ratings["mean"].min() else "#B0B7C3"
          for v in genre_ratings["mean"]]
plt.barh(genre_ratings.index, genre_ratings["mean"], color=colors, edgecolor="black")
plt.gca().invert_yaxis()
plt.xlabel("Average IMDb Rating")
plt.title("Biography Is Undefeated, Horror Can't Catch a Break")
plt.xlim(0, 8)
for i, (v, n) in enumerate(zip(genre_ratings["mean"], genre_ratings["count"])):
    plt.text(v + 0.05, i, f"{v:.2f}  (n={n})", va="center", fontsize=9)
plt.show()
```

Among genres with at least 50 films to their name, **Biography tops the leaderboard at 6.58**,
nearly two full points clear of **Horror at the bottom, on 4.70**. The pattern is a familiar one —
real-life, character-driven stories tend to pull in an audience that rates generously, while horror
draws a bigger, rowdier crowd that's much harder to please.

##  Question 4: Does a Longer Runtime Make a Better Film?

```python
plt.figure(figsize=(7.5, 5.5))
plt.scatter(df["runtime"], df["imdb_rating"], alpha=0.3, color=PALETTE[0], edgecolor="none")
corr = df["runtime"].corr(df["imdb_rating"])
plt.xlabel("Runtime (minutes)")
plt.ylabel("IMDb Rating")
plt.title(f"Runtime Has  Nothing to Say About Quality (r = {corr:.2f})")
plt.show()
```

##  Question 5: Does a Famous Cast Guarantee a Good Movie?


```python
actor_counts = df["actors"].dropna().str.split("|").explode().str.strip()
actor_counts = actor_counts[actor_counts != ""].value_counts()
top_actors = actor_counts.head(10)

plt.figure(figsize=(8, 5))
top_actors.sort_values().plot(kind="barh", color=PALETTE[1], edgecolor="black")
plt.xlabel("Number of Film Appearances")
plt.title("The Ten Busiest Faces in Indian Cinema, 2001-2019")
plt.show()
```

```python
top8_names = top_actors.head(8).index.tolist()
has_top_actor = df["actors"].fillna("").apply(lambda s: any(name in s for name in top8_names))

comparison = pd.DataFrame({
    "Group": ["Films with a top-8 actor", "Films without one"],
    "Count": [has_top_actor.sum(), (~has_top_actor).sum()],
    "Average Rating": [df[has_top_actor]["imdb_rating"].mean(), df[~has_top_actor]["imdb_rating"].mean()],
})

plt.figure(figsize=(6.5, 5))
bars = plt.bar(comparison["Group"], comparison["Average Rating"],
                color=[PALETTE[1], PALETTE[2]], edgecolor="black")
plt.ylabel("Average IMDb Rating")
plt.title("Star Power vs. Quality")
plt.ylim(0, 7)
for bar, val in zip(bars, comparison["Average Rating"]):
    plt.text(bar.get_x() + bar.get_width()/2, val + 0.08, f"{val:.2f}", ha="center", fontweight="bold")
plt.show()
comparison
```

**Anupam Kher tops the list with 90 films** — more than any other actor in the dataset, ahead of
Akshay Kumar (71) and Om Puri (66). 

And here's the twist: films featuring one of these eight busiest actors average **5.46**, while
films *without* any of them average **5.59** — slightly *higher*.

##  Question 6: Do Awards Actually Track With Ratings?

```python
plt.figure(figsize=(7.5, 5.5))
plt.scatter(df["total_recognition"], df["imdb_rating"], alpha=0.35, color=PALETTE[3], edgecolor="none")
plt.xlim(-2, 50)
corr = df["total_recognition"].corr(df["imdb_rating"])
plt.xlabel("Total Wins + Nominations")
plt.ylabel("IMDb Rating")
plt.title(f"Awards and Audience Opinion (r = {corr:.2f})")
plt.show()
```

##  Question 7: Is Indian Cinema Actually Getting Better?

```python
yearly = df.groupby("year_of_release").agg(
    Film_Count=("imdb_rating", "count"),
    Average_Rating=("imdb_rating", "mean"),
)

fig, ax1 = plt.subplots(figsize=(10, 5))
ax1.bar(yearly.index, yearly["Film_Count"], color="#E8E8F0", label="Films Released")
ax1.set_ylabel("Number of Films")
ax1.set_xlabel("Year")

ax2 = ax1.twinx()
ax2.plot(yearly.index, yearly["Average_Rating"], color=PALETTE[0], marker="o", linewidth=2.4, label="Avg Rating")
ax2.set_ylabel("Average IMDb Rating")
ax2.set_ylim(4.8, 6.3)

fig.suptitle("The Class of 2017-2019 Is Rating  Higher", fontweight="bold")
fig.legend(loc="upper left", bbox_to_anchor=(0.12, 0.88))
plt.show()
```

## Question 8: Which Genres Actually Hang Out Together

```python
top6 = ["Drama", "Comedy", "Romance", "Action", "Crime", "Thriller"]
co_matrix = pd.DataFrame(0, index=top6, columns=top6)
for gl in df["genres_list"]:
    present = [g for g in gl if g in top6]
    for i in present:
        for j in present:
            co_matrix.loc[i, j] += 1

plt.figure(figsize=(7.5, 6.5))
sns.heatmap(co_matrix, annot=True, fmt="d", cmap="rocket_r", square=True,
            linewidths=0.5, cbar_kws={"label": "Films sharing both genres"})
plt.title(" Who Shows Up Together")
plt.show()
```

**Comedy and Romance pair up in 189 films** — the rom-com formula, alive and well — while
**Action and Crime team up 140 times**, the cops-and-gangsters staple. **Crime and Thriller**, despite
sounding like a natural match, overlap in only 61 films .

##  The 2000s vs. the 2010s

```python
decade_stats = df.groupby("decade").agg(
    Films=("imdb_rating", "count"),
    Avg_Rating=("imdb_rating", "mean"),
    Avg_Runtime=("runtime", "mean"),
).rename(index={2000: "2000s", 2010: "2010s"})

fig, ax = plt.subplots(figsize=(6.5, 5))
bars = ax.bar(decade_stats.index, decade_stats["Avg_Rating"], color=[PALETTE[3], PALETTE[0]], edgecolor="black", width=0.5)
ax.set_ylim(0, 7)
ax.set_ylabel("Average IMDb Rating")
ax.set_title("The 2010s Edged Out the 2000s")
for bar, val, n in zip(bars, decade_stats["Avg_Rating"], decade_stats["Films"]):
    ax.text(bar.get_x() + bar.get_width()/2, val + 0.1, f"{val:.2f}\n({n} films)", ha="center", fontweight="bold")
plt.show()
```

# Question 9:How Every Number in This Dataset Relates to Each Other


```python
numeric_cols = ["imdb_rating", "imdb_votes", "runtime", "wins", "nominations", "total_recognition"]
corr_matrix = df[numeric_cols].corr()

plt.figure(figsize=(7.5, 6.5))
sns.heatmap(corr_matrix, annot=True, fmt=".2f", cmap="coolwarm", vmin=-1, vmax=1,
            square=True, linewidths=0.5, cbar_kws={"label": "Correlation"})
plt.title("How Every Number in This Dataset Relates to Each Other")
plt.tight_layout()
plt.show()
```

## Conclusiom

A familiar cast, a longer runtime, a bigger opening — none of
them reliably predict quality on their own. What actually shows up in the data is smaller and more
specific: Biography quietly outperforms, Romance quietly declined, and the loudest, most-voted films
aren't always the best-loved ones.


- A 0 rating almost certainly means "not enough votes yet," not "audiences hated it" — left in, but
  worth remembering.
- `wins_nominations` gaps are treated as "nothing recorded," not a confirmed zero.
- `actors` doesn't separate lead roles from supporting ones, so the star-power finding describes
  overall cast presence, not lead-performance quality specifically.
