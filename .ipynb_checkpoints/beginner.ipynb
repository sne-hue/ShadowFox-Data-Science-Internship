## 1. Introduction

Visualization means showing data as a picture instead of as rows and numbers. A table of 950 IPL
matches tells us very little when we just look at it. The same data as a bar chart tells us in one
second which team has won the most matches.

Python has many libraries for this — Matplotlib, Seaborn, Plotly, Bokeh and Pandas. For this task I
have picked **Pandas** and **Matplotlib**.

I picked these two because they are connected. Pandas plotting is actually built on top of
Matplotlib, so learning them together shows both the short way and the long way of making the same
chart. Pandas gives one line of code and a quick chart. Matplotlib gives more code but full control
over how the chart looks.

Every example in this guide uses a real IPL dataset instead of random numbers, so each chart
actually answers a cricket question.

## 2. About the dataset

The dataset has one row per IPL match played between 2008 and 2022.

| Column | Meaning |
|---|---|
| ID | Unique match id |
| City, Venue | Where the match was played |
| Date, Season | When the match was played |
| MatchNumber | League match number, or Final / Qualifier / Eliminator |
| Team1, Team2 | The two teams |
| TossWinner, TossDecision | Who won the toss and whether they chose to bat or field |
| WinningTeam, WonBy, Margin | Result of the match |
| Player_of_Match | Man of the match |
| Umpire1, Umpire2 | On field umpires |

### Importing the libraries

`pandas` is for loading and shaping the data. `matplotlib.pyplot` is the plotting part of
Matplotlib.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r"C:\Users\User\Desktop\shadowfox\ipl-matches (1).csv")

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
df.head()
```

```python
df.isnull().sum()
```

```python
df.info()
```


**Description:**
`read_csv()` reads the file into a DataFrame. `shape` gives (rows, columns) and `head()` shows the
first five rows so we can check the file loaded correctly.


`info()` shows the column names and the data type.
`isnull().sum()` counts the missing values column by column.


### Cleaning the data


```python
# Season is stored as text and some seasons are written like "2007/08".
season_fix = {"2007/08": 2008, "2009/10": 2010, "2020/21": 2020}
df["Year"] = df["Season"].replace(season_fix).astype(int)
results = df.dropna(subset=["WinningTeam"]).copy()
print("Seasons covered:", df["Year"].min(), "to", df["Year"].max())
print("Matches with a result:", len(results))
results[["Year", "Team1", "Team2", "WinningTeam", "WonBy", "Margin"]].head()
```

# PART 1 — PANDAS VISUALIZATION

Pandas is an open-source Python library used for data manipulation, data analysis, and data cleaning. It provides easy-to-use data structures such as Series and DataFrame, which make it convenient to work with structured data.

### Key Features of Pandas

**DataFrame:** Represents data in a table-like format with rows and columns.

**Data Cleaning:** Helps handle missing values, duplicates, and inconsistent data.

**Data Selection:** Allows specific rows and columns to be selected easily.

**Data Filtering:** Makes it easy to find records that satisfy particular conditions.

**Data Aggregation:** Functions such as `groupby()` can summarize data based on categories.

**Data Import and Export:** Supports formats such as CSV and Excel.

**Statistical Analysis:** Provides functions for calculating measures such as mean, count, minimum, maximum, and standard deviation.

 
### 3.1 LINE PLOT
**how many matches were played each season?**

```python

matches_per_season = df["Year"].value_counts().sort_index()

matches_per_season.plot(kind="line", marker="o", color="darkgreen",
                        title="Matches Played in Each IPL Season")
plt.xlabel("Season")
plt.ylabel("Number of Matches")
plt.grid(True)
plt.show()
```

### 3.2 BAR CHART
 **which team has won the most matches?**

```python
wins = results["WinningTeam"].value_counts().head(10)

wins.plot(kind="bar", color="lightblue", edgecolor="black",
          title="Top 10 Teams by Total Wins (2008-2022)",rot=75)
plt.xlabel("Team")
plt.ylabel("Matches Won")
plt.show()
```

### 3.3 HORIZONTAL BAR CHART
**who has won the most Player of the Match awards?**

```python
pma = df["Player_of_Match"].value_counts().head(10).sort_values()

pma.plot(kind="barh", color="indianred",
          title="Top 10 Players by Player of the Match Awards")
plt.xlabel("Number of Awards")
plt.ylabel("Player")
plt.show()
```

### 3.4 HISTOGRAM
 **by how many runs are matches usually won?**

```python
runs_margin = results[results["WonBy"] == "Runs"]["Margin"]

runs_margin.plot(kind="hist", bins=20, color="mediumpurple", edgecolor="black",
                 title="Distribution of Victory Margin (Runs)")
plt.xlabel("Runs")
plt.ylabel("Number of Matches")

plt.show()
```

### 3.5 DENSITY (KDE) PLOT

A KDE plot is a smooth version of a histogram. Instead of bars it draws one curve estimating where
the values are concentrated. It is useful for comparing two distributions on the same axes.

```python
runs_margin.plot(kind="kde", color="darkorange", linewidth=2,
                 title="Density of Victory Margin (Runs)")
plt.xlabel("Runs")
plt.grid(True)
plt.show()
```

### 3.6 BOX PLOT
 **how do run margins and wicket margins compare in spread?**

```python
margins = pd.DataFrame({
    "Won by Runs": results[results["WonBy"] == "Runs"]["Margin"].reset_index(drop=True),
    "Won by Wickets": results[results["WonBy"] == "Wickets"]["Margin"].reset_index(drop=True)
})

margins.plot(kind="box", color="green",
             title="Spread of Victory Margins")
plt.ylabel("Margin")
plt.grid(True)
plt.show()
```

### 3.7 PIE CHART

 **after winning the toss, do captains bat or field?**

```python
toss = df["TossDecision"].value_counts()

toss.plot(kind="pie", autopct="%1.1f%%", startangle=90,
          colors=["#6051B5", "#52cd3f"],
          wedgeprops={"edgecolor": "white", "linewidth": 2},
          title="Toss Decision Across All Seasons")
plt.ylabel("")
plt.show()
```

### 3.8 AREA CHART

An area chart is a line chart with the space under the line filled in. The filled area makes the
total volume easier to feel than a plain line.

```python
matches_per_season.plot(kind="area", color="teal", alpha=0.4,
                        title="Matches Played per Season (Area)")
plt.xlabel("Season")
plt.ylabel("Number of Matches")
plt.show()
```

### 3.9 STACKED BAR CHART

**has the toss decision changed over the years?**

```python
toss_by_year = pd.crosstab(df["Year"], df["TossDecision"])

toss_by_year.plot(kind="bar", stacked=True,
                  color=["#6cd959", "#88d2ca"],
                  title="Toss Decision by Season", rot=45)
plt.xlabel("Season")
plt.ylabel("Number of Matches")
plt.legend(title="Decision")
plt.show()
```

### 3.10 SCATTER PLOT
**do teams that play more matches also win more?**

```python
played = pd.concat([df["Team1"], df["Team2"]]).value_counts()
won = results["WinningTeam"].value_counts()

team_stats = pd.DataFrame({"Played": played, "Won": won}).dropna()

team_stats.plot(kind="scatter", x="Played", y="Won", s=70, color="crimson",
                title="Matches Played vs Matches Won (per team)")
plt.xlabel("Matches Played")
plt.ylabel("Matches Won")
plt.show()

team_stats.sort_values("Won", ascending=False).head()
```

# PART 2 — MATPLOTLIB VISUALIZATION

## 4. Matplotlib overview

Matplotlib was created by John Hunter in 2003 and is the base of almost all Python plotting. Pandas,
Seaborn and several others are built on it.

A Matplotlib chart is made of four parts:

**Figure** — the whole canvas. One figure can hold one plot or many.

**Axes** — a single plot inside the figure. This is where the data is drawn. `set_title()`,
`set_xlabel()` and `set_ylabel()` belong to the axes.

**Axis** — the x line and y line, which control the range and the tick marks.

**Artist** — every single thing drawn on the figure: lines, bars, text, legend. Everything visible
is an artist.


### 4.1 LINE PLOT

`plot()` is the default function of Matplotlib. Compared to the pandas version, here the x and y
values have to be passed separately — which is more work, but it also means the two can come from
anywhere.

```python
x = matches_per_season.index
y = matches_per_season.values

plt.figure(figsize=(9, 4.5))
plt.plot(x, y, color="darkgreen", marker="o", linestyle="--", linewidth=2, markersize=7)
plt.title("Matches Played in Each IPL Season", fontsize=13, fontweight="bold")
plt.xlabel("Season")
plt.ylabel("Number of Matches")
plt.xticks(x, rotation=45)
plt.show()
```

### 4.2 BAR CHART

`bar()` takes the category names and the heights. Unlike pandas it does not sort or count anything —
the data has to be ready before plotting.
 **which grounds have hosted the most matches?**

```python
venues = df["Venue"].value_counts().head(8)
short_names = [v.split(",")[0] for v in venues.index]

plt.figure(figsize=(9, 5))
bars = plt.bar(short_names, venues.values, color="#46b4ab", width=0.6, edgecolor="black")

for bar in bars:
    plt.text(bar.get_x() + bar.get_width() / 2, bar.get_height() + 1,
             int(bar.get_height()), ha="center", fontsize=9)

plt.title("Top 8 Venues by Matches Hosted", fontsize=13, fontweight="bold")
plt.xlabel("Venue")
plt.ylabel("Matches Hosted")
plt.xticks(rotation=70)
plt.tight_layout()
plt.show()
```

### 4.3 GROUPED BAR CHART

 **how many of the matches it played did each big team win?**

```python
top6 = team_stats.sort_values("Won", ascending=False).head(6)
labels = [name.replace(" ", "\n") for name in top6.index]
positions = range(len(top6))
width = 0.4

plt.figure(figsize=(10, 5))
plt.bar([p - width/2 for p in positions], top6["Played"], width,
        label="Played", color="lightsteelblue", edgecolor="black")
plt.bar([p + width/2 for p in positions], top6["Won"], width,
        label="Won", color="#b5b952", edgecolor="black")

plt.xticks(list(positions), labels, fontsize=9)
plt.title("Matches Played vs Won - Top 6 Teams", fontsize=13, fontweight="bold")
plt.ylabel("Number of Matches")
plt.legend()
plt.tight_layout()
plt.show()
```

### 4.4 HISTOGRAM


```python
early = results[(results["Year"] <= 2014) & (results["WonBy"] == "Runs")]["Margin"]
recent = results[(results["Year"] > 2014) & (results["WonBy"] == "Runs")]["Margin"]

plt.figure(figsize=(9, 4.5))
plt.hist(early, bins=15, alpha=0.6, label="2008-2014", color="purple", edgecolor="white")
plt.hist(recent, bins=15, alpha=0.6, label="2015-2022", color="darkorange", edgecolor="white")
plt.title("Victory Margin in Runs - Early vs Recent Seasons", fontsize=13, fontweight="bold")
plt.xlabel("Runs")
plt.ylabel("Number of Matches")
plt.legend()
plt.show()
```

### 4.5 SCATTER PLOT WITH ANNOTATION

`scatter()` gives more control than the pandas version: the colour and the size of every dot can
come from the data.

```python
team_stats["win"] = team_stats["Won"] / team_stats["Played"] * 100
big_teams = team_stats[team_stats["Played"] >= 100]

plt.figure(figsize=(9, 5.5))
sc = plt.scatter(big_teams["Played"], big_teams["Won"],
                 s=big_teams["win"] * 6, c=big_teams["win"],
                 cmap="viridis", edgecolor="black", alpha=0.85)

for team in big_teams.index:
    plt.annotate(team.split()[0], (big_teams.loc[team, "Played"], big_teams.loc[team, "Won"]),
                 textcoords="offset points", xytext=(6, 6), fontsize=8)

plt.colorbar(sc, label="Win %")
plt.title("Matches Played vs Won (teams with 100+ matches)", fontsize=13, fontweight="bold")
plt.xlabel("Matches Played")
plt.ylabel("Matches Won")
plt.tight_layout()
plt.show()
```

### 4.6 PIE CHART

`pie()` with `explode` pulls one slice out to highlight it.
 **how are IPL matches decided?**

```python
won_by = results["WonBy"].value_counts()

plt.figure(figsize=(6.5, 6.5))
plt.pie(won_by.values, labels=won_by.index, autopct="%1.1f%%", startangle=140,
        explode=[0, 0, 0.15, 0.25][:len(won_by)],
        colors=["#cee96d", "#21b400", "#ba5671", "#c9a34a"],
        wedgeprops={"edgecolor": "white", "linewidth": 2},
        textprops={"fontsize": 10})
plt.title("How IPL Matches Were Decided", fontsize=13, fontweight="bold")
plt.show()
```

### 4.7 AREA / STACKED AREA CHART



```python
plt.figure(figsize=(9.5, 5))
plt.stackplot(toss_by_year.index,
              toss_by_year["bat"], toss_by_year["field"],
              labels=["Chose to bat", "Chose to field"],
              colors=["#cd853f", "#2e8b57"], alpha=0.8)
plt.title("Toss Decisions Over the Seasons", fontsize=13, fontweight="bold")
plt.xlabel("Season")
plt.ylabel("Number of Matches")
plt.legend(loc="upper left")
plt.grid(True)
plt.show()
```

### 4.8 BOX PLOT



```python
data = [results[results["WonBy"] == "Runs"]["Margin"].dropna(),
        results[results["WonBy"] == "Wickets"]["Margin"].dropna()]

plt.figure(figsize=(7.5, 5))
box = plt.boxplot(data, tick_labels=["Won by Runs", "Won by Wickets"],
                  patch_artist=True, medianprops={"color": "black", "linewidth": 2})

for patch, colour in zip(box["boxes"], ["#4682b4", "#b22222"]):
    patch.set_facecolor(colour)
    patch.set_alpha(0.7)

plt.title("Spread of Victory Margins", fontsize=13, fontweight="bold")
plt.ylabel("Margin")
plt.grid(True)
plt.show()
```

### 4.9 SUBPLOTS

`subplots()` creates a grid of plots inside one figure. 

```python
fig, ax = plt.subplots(2, 2, figsize=(12, 8))

ax[0, 0].plot(x, y, color="darkgreen", marker="o")
ax[0, 0].set_title("Matches per Season")
ax[0, 0].set_xlabel("Season")

ax[0, 1].bar(wins.head(5).index.str.split().str[0], wins.head(5).values, color="steelblue")
ax[0, 1].set_title("Top 5 Teams by Wins")

ax[1, 0].hist(runs_margin, bins=18, color="mediumpurple", edgecolor="white")
ax[1, 0].set_title("Victory Margin (Runs)")
ax[1, 0].set_xlabel("Runs")

ax[1, 1].pie(toss.values, labels=toss.index, autopct="%1.0f%%",
             colors=["#2e8b57", "#cd853f"])
ax[1, 1].set_title("Toss Decision")

fig.suptitle("IPL 2008-2022 Summary Dashboard", fontsize=15, fontweight="bold")
plt.tight_layout()
plt.show()
```

### 4.10 CUSTOMISATION AND SAVING


```python
plt.figure(figsize=(9.5, 5))
plt.plot(x, y, color="#b22222", marker="s", linewidth=2, label="Matches played")

average = y.mean()
plt.axhline(average, color="grey", linestyle=":", linewidth=2,
            label=f"Average = {average:.0f}")

peak_year = matches_per_season.idxmax()
peak_value = matches_per_season.max()
plt.annotate(f"Highest: {peak_value} matches",
             xy=(peak_year, peak_value), xytext=(peak_year - 5, peak_value + 3),
             arrowprops={"arrowstyle": "->", "color": "black"}, fontsize=10)

plt.ylim(40, 85)
plt.title("IPL Matches per Season with Average Line", fontsize=13, fontweight="bold")
plt.xlabel("Season")
plt.ylabel("Number of Matches")
plt.legend()
plt.tight_layout()
plt.savefig("matches_per_season.png", dpi=150)
plt.show()
```

## 5. Comparison of Pandas and Matplotlib

| Point | Pandas `.plot()` | Matplotlib |
|---|---|---|
| Lines of code | One line for a finished chart | Several lines for the same chart |
| Input format | Needs the data already grouped in a DataFrame | Takes plain lists, arrays or Series |
| Labels | Index and column names become axis labels and legend automatically | Every label must be written by hand |
| Chart types | About 10, chosen with `kind=` | Everything, including 3D, maps and animations |
| Customisation | Limited; anything extra falls back to Matplotlib | Complete control over every element |
| Multiple plots | Awkward | `subplots()` handles it cleanly |
| Interactivity | None (static images) | Static by default; zoom and pan only with widget backends |
| Large datasets | Same speed as Matplotlib, since it uses Matplotlib | Fast for normal sizes, slows down past a few lakh points |

**Strengths of Pandas plotting:** speed of writing. During exploration, when a chart is going to be
looked at once and thrown away, `df.plot(kind="bar")` is hard to beat. It also keeps the code close
to the data cleaning, so there is no jumping between libraries.

**Weaknesses of Pandas plotting:** the moment something unusual is needed — a value written on top of
a bar, two charts side by side, an arrow pointing at a spike — it runs out. The data also has to be
in exactly the right shape first, which sometimes takes more work than the plot itself.

**Strengths of Matplotlib:** control. Every colour, tick, label, line and piece of text can be
changed. It also works with any input, not only DataFrames, and is the only one of the two that can
build a multi-panel figure like the dashboard in section 4.9.

**Weaknesses of Matplotlib:** it is verbose. The grouped bar chart in section 4.3 needed manual
position shifting for something that should be one option. Its default styling also looks plain and
usually needs work before it can be shown to anyone.



## 7. Conclusion

Working through both libraries on the same dataset made the difference between them obvious. Pandas
is the faster way to look at data, Matplotlib is the better way to present it. Since pandas plotting
is Matplotlib underneath, learning Matplotlib is not optional — it is what lets you fix anything
pandas cannot do on its own.
