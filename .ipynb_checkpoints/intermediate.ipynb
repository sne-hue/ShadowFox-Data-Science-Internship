## 2. Converting readings into an AQI

India's AQI (via CPCB) isn't just the raw PM2.5 number — it's calculated from a formula that maps
the pollutant concentration onto a 0-500 scale, using different breakpoint tables for each pollutant.
The worst sub-index among all pollutants becomes the AQI for that reading.

Since PM2.5 and PM10 are what drive Delhi's air quality problem in winter (the other gases matter but
rarely push the index higher than particulate matter does), I'm computing the AQI using just those
two, which is a reasonable simplification for this analysis.

The CPCB breakpoint tables look like this (concentration in µg/m3 → sub-index):

```python
# CPCB breakpoints: (conc_low, conc_high, index_low, index_high)
PM25_BREAKPOINTS = [
    (0, 30, 0, 50), (31, 60, 51, 100), (61, 90, 101, 200),
    (91, 120, 201, 300), (121, 250, 301, 400), (251, 350, 401, 500),
]
PM10_BREAKPOINTS = [
    (0, 50, 0, 50), (51, 100, 51, 100), (101, 250, 101, 200),
    (251, 350, 201, 300), (351, 430, 301, 400), (431, 510, 401, 500),
]

def sub_index(cp, table):
    for c_lo, c_hi, i_lo, i_hi in table:
        if c_lo <= cp <= c_hi:
            return ((i_hi - i_lo) / (c_hi - c_lo)) * (cp - c_lo) + i_lo
    return 500  
df["pm25_index"] = df["pm2_5"].apply(lambda x: sub_index(x, PM25_BREAKPOINTS))
df["pm10_index"] = df["pm10"].apply(lambda x: sub_index(x, PM10_BREAKPOINTS))
df["AQI"] = df[["pm25_index", "pm10_index"]].max(axis=1)

def categorise(aqi):
    if aqi <= 50: return "Good"
    if aqi <= 100: return "Satisfactory"
    if aqi <= 200: return "Moderate"
    if aqi <= 300: return "Poor"
    if aqi <= 400: return "Very Poor"
    return "Severe"

df["Category"] = df["AQI"].apply(categorise)
df[["date", "pm2_5", "pm10", "AQI", "Category"]].head()
```

### 2.1 How many hours fell into each category

```python
counts = df["Category"].value_counts()
order = ["Good", "Satisfactory", "Moderate", "Poor", "Very Poor", "Severe"]
counts = counts.reindex(order).dropna()
colors = {"Good": "#2ecc71", "Satisfactory": "#a3d977", "Moderate": "#f5d547",
          "Poor": "#f39c12", "Very Poor": "#e74c3c", "Severe": "#7b241c"}

plt.figure(figsize=(8, 5))
plt.bar(counts.index, counts.values, color=[colors[c] for c in counts.index], edgecolor="black")
plt.title("Hours Spent in Each AQI Category — Jan 1-24, 2023")
plt.ylabel("Number of Hours")
for i, v in enumerate(counts.values):
    plt.text(i, v + 5, f"{v} hrs\n({v/len(df)*100:.0f}%)", ha="center", fontsize=9)
plt.show()
```

**Severe** and **Very Poor** together account for
roughly 97% of all the hours in this window — barely any time was spent in Moderate or better.

## 3. Does pollution follow a pattern through the day?

Traffic, cooking fires, and temperature inversion all follow a daily rhythm, so I'd expect pollution
to as well. I grouped every reading by the hour of the day (0 to 23) regardless of which date it fell
on, and averaged PM2.5 across all 24 days for each hour.

```python
hourly_avg = df.groupby(df["date"].dt.hour)[["pm2_5", "pm10"]].mean()

plt.figure(figsize=(10, 5))
plt.plot(hourly_avg.index, hourly_avg["pm2_5"], marker="o", color="#189435", label="PM2.5")
plt.plot(hourly_avg.index, hourly_avg["pm10"], marker="o", color="#2a61c8", label="PM10")
plt.title("Average Pollutant Level by Hour of Day (averaged over 24 days)")
plt.xlabel("Hour (24-hr)")
plt.ylabel("Concentration (µg/m³)")
plt.xticks(range(0, 24, 2))
plt.legend()
plt.show()
```

There is a clear pattern. I assumed the classic "morning rush
hour spike" story, but the lowest point of the whole day is actually around **8-9 AM**, and the levels
climb steadily through the afternoon to peak around **5-6 PM**, staying high into the evening before
slowly dropping overnight.

The morning sun starts breaking up the inversion layer that trapped overnight pollution, which is why 8-9
AM briefly looks better. Then as evening traffic builds up and the air cools again, the inversion
layer reforms closer to the ground and traps everything that's been emitted through the day, right
around rush hour.

To see this a different way —  here's a boxplot of PM2.5 by hour instead of just the mean line.

```python
df["hour"] = df["date"].dt.hour

plt.figure(figsize=(13, 5.5))
sns.boxplot(data=df, x="hour", y="pm2_5", color="#e67e22")
plt.title("Spread of PM2.5 Readings by Hour of Day")
plt.xlabel("Hour (24-hr)")
plt.ylabel("PM2.5 (µg/m³)")
plt.show()
```

## 4. Which pollutants move together

If several pollutants rise and fall together,  —
mainly vehicle exhaust and burning, in Delhi's case.

```python
pollutants = ["co", "no", "no2", "o3", "so2", "pm2_5", "pm10", "nh3"]
corr = df[pollutants].corr()

plt.figure(figsize=(8, 6.5))
sns.heatmap(corr, annot=True, fmt=".2f", cmap="RdYlGn_r", vmin=-1, vmax=1,
            linewidths=0.5, square=True)
plt.title("Correlation Between Pollutants")
plt.show()
```

A few things jump out:

- **PM2.5 and PM10 move almost in lockstep**  which makes sense since PM10
  includes PM2.5 — the finer particles .
- **CO correlates very strongly with both PM2.5 (0.95) and PM10 (0.97)**. CO is a  marker of
  combustion — vehicles, and this time of year, crop residue burning and open fires used for warmth.
- **Ozone (O3) is the outlier** — it's the only pollutant negatively correlated with almost everything
  else (-0.45 with PM2.5).
- **NH3 correlates most with SO2 (0.84)**, more than with PM2.5 — ammonia mixing with sulphur
  compounds is a known factor in secondary particulate formation.

### 4.1 PM2.5 vs PM10 relationship directly

```python
plt.figure(figsize=(7, 6))
plt.scatter(df["pm2_5"], df["pm10"], alpha=0.4, color="#c0392b", edgecolor="none")
z = np.polyfit(df["pm2_5"], df["pm10"], 1)
x_line = np.linspace(df["pm2_5"].min(), df["pm2_5"].max(), 100)
plt.plot(x_line, z[0]*x_line + z[1], color="black", linewidth=2,
         label=f"PM10 ≈ {z[0]:.2f} × PM2.5 + {z[1]:.1f}")
plt.title("PM2.5 vs PM10")
plt.xlabel("PM2.5 (µg/m³)")
plt.ylabel("PM10 (µg/m³)")
plt.legend()
plt.show()
```

## 5. Best and worst days

```python
daily_avg = df.groupby(df["date"].dt.date)["pm2_5"].mean().sort_values()

plt.figure(figsize=(11, 5))
bar_colors = ["#2ecc71" if v == daily_avg.min() else
              "#7b241c" if v == daily_avg.max() else "#e67e22" for v in daily_avg.values]
plt.bar(daily_avg.index.astype(str), daily_avg.values, color=bar_colors, edgecolor="black")
plt.title("Average Daily PM2.5 — Jan 1-24, 2023")
plt.ylabel("PM2.5 (µg/m³)")
plt.xticks(rotation=60)
plt.axhline(60, color="black", linestyle="--", linewidth=1, label="WHO 24-hr guideline (60 µg/m³)")
plt.legend()
plt.tight_layout()
plt.show()

print("Cleanest day:", daily_avg.idxmin(), "-", round(daily_avg.min(), 1), "µg/m3")
print("Worst day:   ", daily_avg.idxmax(), "-", round(daily_avg.max(), 1), "µg/m3")
```

Every single day in this dataset sits above the WHO guideline line — some of them more than ten times
over. **15 January was the cleanest** day in the whole window at about 128 µg/m3, which even then is
double the WHO limit. **2 January was the worst**, averaging close to 700 µg/m3 — over eleven times
the guideline. 

## 6. Conclusion


1. **Almost the entire window (97% of hours) was in the Severe or Very Poor category.** This isn't a
   dataset with a few bad spikes — bad air was the default state, not the exception.
2. **Pollution has a clear daily rhythm**: lowest around 8-9 AM, climbing through the day to peak
   around 5-6 PM, and staying elevated into the night. 
3. **PM2.5, PM10 and CO rise and fall together**, which points toward combustion (traffic and burning)
   as the dominant source in this window, rather than something like construction dust. 
4. **Even the cleanest day was double the WHO safe limit**, and the worst day was over eleven times
   the limit.


**Limitations worth being upfront about:**
 Delhi's pollution is known to vary a lot by area. The AQI here is also
